```python
import numpy as np
```

```python
features = np.array([2.0, 3.0])
weights = np.array([4.0, -1.0])
bias = 5.0

def weighted_predict(features, weights, bias):
    if not features.shape[0] == weights.shape[0]:
        raise ValueError("Incompatible dimensions")

    result = np.dot(features, weights) + bias
    return result

print(weighted_predict(features, weights, bias))
```

```python
features = np.array([2.0, 3.0, 4.0])
weights = np.array([4.0, -1.0])
bias = 5.0

weighted_predict(features, weights, bias)
```

```python
features = np.array([
    [2.0, 3.0],
    [4.0, 1.0],
    [1.0, 5.0]
])

weights = np.array([4.0, -1.0])
bias = 5.0

def predict_batch(features_matrix, weights, bias):
    if not features_matrix.shape[1] == weights.shape[0]:
        raise ValueError("Incompatible Dimensions")

    result = np.dot(features_matrix, weights) + bias
    return result

auto_result = predict_batch(features, weights, bias)
manual_result = np.array([10, 20, 4])

assert np.allclose(manual_result, auto_result)
print("Success")
```

```python
features = np.array([
    [2.0, 3.0, 4.0],
    [4.0, 1.0, 0.0]
])

weights = np.array([4.0, -1.0])
bias = 5.0
try: 
    predict_batch(features, weights, bias)
except ValueError:
    print("Test passed: incompatible dimensions rejected")
else:
    raise AssertionError("Function returns prediction.")
```

```python
features = np.array([2.0, 3.0])
weights = np.array([4.0, -1.0])
bias = 5.0

func_result = weighted_predict(features, weights, bias)
man_result = 10.0

assert np.isclose(man_result, func_result)
print("Single prediction test passed")
```
