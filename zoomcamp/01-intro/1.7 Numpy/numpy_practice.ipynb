```python
import numpy as np
```

```python
array1 = np.arange(1, 6)
print(array1)
print(type(array1))
print(array1.shape)

array2 = np.arange(1, 5) * 10
print(array2)
print(type(array2))
print(array2.shape)

array3 = np.arange(-2, 6, 2)
print(array3)
print(type(array3))
print(array3.shape)
```

```python
test1 = [1, 2, 3, 4, 5]
test2 = [10, 20, 30, 40]
test3 = [-2, 0, 2, 4]

def mean_python(numbers):
    mean = 0
    for n in numbers:
        mean += n
    mean = mean / len(numbers)
    return mean

print(mean_python(array1))
print(mean_python(test1))
print(mean_python(test2))
print(mean_python(test3))
```

```python
def mean_numpy(numbers):
    return np.mean(numbers)

print(mean_python(test1) == mean_numpy(test1))
print(mean_python(test2) == mean_numpy(test2))
print(mean_python(test3) == mean_numpy(test3))
```

```python
t1 = [1, 2, 3]
t2 = [2, 2, 2, 2]

def variance_python(numbers):
    total = 0
    mean = mean_python(numbers)
    for n in numbers:
        total += (n - mean) ** 2
    variance = total / len(numbers)
    return variance

print(variance_python(t1))
print(variance_python(t2))
```

```python
t1 = [1, 2, 3]
t2 = [2, 2, 2, 2]
t3 = [-2, 0, 2, 4]

def variance_numpy(numbers):
    variance = np.var(numbers)
    return variance

print(variance_python(t1) == variance_numpy(t1))
print(variance_python(t2) == variance_numpy(t2))
print(variance_python(t3) == variance_numpy(t3))
```

```python
a = [1, 2, 3]
b = [4, 5, 6]

def dot_python(a, b):
    total = 0
    if len(a) == len(b):
        for i in range(len(a)):
            total += a[i] * b[i]

        return total
    else:
        raise ValueError("The two vectors must have the same length.")

print(dot_python([1, 2, 3], [4, 5, 6]))
print(dot_python([2, 4, 6], [3, 1, 2]))
print(dot_python([1, 2, 3], [4, 5]))
```

```python
d1_a = [1, 2, 3]
d1_b = [4, 5, 6]

d2_a = [2, 4, 6]
d2_b = [3, 1, 2]

d3_a = [-1, 0, 2]
d3_b = [3, 5, 4]

def dot_numpy(a, b):
    return np.dot(a, b)

print(dot_numpy(d1_a, d1_b))
print(dot_numpy(d2_a, d2_b))
print(dot_numpy(d3_a, d3_b))
```

```python
print(np.isclose(dot_numpy(d1_a, d1_b), dot_python(d1_a, d1_b)))
print(np.isclose(dot_numpy(d2_a, d2_b), dot_python(d2_a, d2_b)))
print(np.isclose(dot_numpy(d3_a, d3_b), dot_python(d3_a, d3_b)))
```
