```python
import pandas as pd
import numpy as np

def predict_linear_loops(X, weights, bias):
    pred = []

    for i in range(len(X)):
        row_pred = bias
        if len(X[i]) != len(weights):
            raise ValueError("Number of features in X does not match number of weights.")
        for j in range(len(X[i])):
            row_pred += X[i][j] * weights[j]
        pred.append(row_pred)

    return pred

X = [
    [2, 3],
    [1, 4]
]

weights = [4, -1]
bias = 5

predictions = predict_linear_loops(X, weights, bias)

print(predictions)
```

```python
def predict_linear_numpy(X, weights, bias):
    pred = np.dot(X, weights) + bias
    return pred
```

```python
X = np.array([
    [2, 3],
    [1, 4]
])

weights = np.array([4, -1])
bias = 5

predictions = predict_linear_numpy(X, weights, bias)

print(predictions)
print(predictions.shape)
```

```python
X = [
    [2, 1, 3],
    [0, 4, 2]
]

weights = [3, -2, 5]
bias = -4
```

```python
loop_predictions = predict_linear_loops(X, weights, bias)

numpy_predictions = predict_linear_numpy(
    np.array(X),
    np.array(weights),
    bias
)

print("Loops:", loop_predictions)
print("NumPy:", numpy_predictions)

assert np.allclose(loop_predictions, numpy_predictions)
```

```python
def test_predict_linear():
    test_cases = [
        # Test Case 1: Basic functionality
        (
            [[2, 3], [1, 4]],
            [4, -1],
            5,
            [10, 5]
        ),

        # Test Case 2: Three features with negative bias
        (
            [[2, 1, 3], [0, 4, 2]],
            [3, -2, 5],
            -4,
            [15, -2]
        ),

        # Test Case 3: Floating-point values and zero weights
        (
            [[1.5, 2.0, 3.0], [0.5, -1.0, 4.0]],
            [2.0, 0.0, -0.5],
            1.5,
            [3.0, 0.5]
        ),

        # Test Case 4: Single feature and zero bias
        (
            [[0], [-2], [3.5]],
            [-4],
            0,
            [0, 8, -14]
        ),

        # Test Case 5: Four features with mixed signs
        (
            [[1, 0, 2, -1], [-2, 3, 0, 4]],
            [2, -3, 0.5, 1],
            -2,
            [0, -11]
        )
    ]

    for i, (X, weights, bias, expected) in enumerate(test_cases, start=1):
        loop_predictions = predict_linear_loops(X, weights, bias)
        numpy_predictions = predict_linear_numpy(X, weights, bias)

        # Verify both against independently calculated values
        assert np.allclose(loop_predictions, expected), \
            f"Test Case {i}: Loop implementation failed"

        assert np.allclose(numpy_predictions, expected), \
            f"Test Case {i}: NumPy implementation failed"

        # Verify both implementations agree
        assert np.allclose(loop_predictions, numpy_predictions), \
            f"Test Case {i}: Implementations disagree"

        print(f"Test Case {i}: PASS")

    print("\nAll 5 test cases passed!")


# Run all tests
test_predict_linear()
```
